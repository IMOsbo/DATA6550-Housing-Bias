# Preprocess Ramsey County

After doing some digging around, I found a layer with all of the coventants!

<https://services.arcgis.com/8df8p0NlLFEShl0r/arcgis/rest/services/All_Covenants_8_4_2018/FeatureServer/0>

It's an ArcGIS Feature Server, so I figured out how to query it and get all of the features. 

<https://services.arcgis.com/8df8p0NlLFEShl0r/arcgis/rest/services/All_Covenants_8_4_2018/FeatureServer/0/query?where=1=1&outFields=*&f=json>

(Thanks ChatGPT for helping me figure that one out!)

```python
import requests
import pandas as pd
```

## Download the JSON

```python
response = requests.get("https://services.arcgis.com/8df8p0NlLFEShl0r/arcgis/rest/services/All_Covenants_8_4_2018/FeatureServer/0/query?where=1=1&outFields=*&f=json")
```

Now let's convert it to a dictionary...

```python
data = response.json()["features"]
data[0]
```

Ok, now it's a list of dictionaries. We're interested in only the attributes part, so let's loop over `data` and pull out that instead.

```python
cleanedData = []
for covenant in data:
    cleanedData.append(covenant["attributes"])
```

```python
len(cleanedData)
```

## Convert to Pandas and save as CSV

```python
ramseyData = pd.DataFrame(cleanedData)
ramseyData
```

```python
ramseyData.to_csv("ramsey_Covenants.csv", index=False)
```
